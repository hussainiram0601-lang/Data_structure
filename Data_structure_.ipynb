```python
#BFS
# Using queue
class Queue:
  def __init__(self):
    self.q = []
    self.front = -1

  def push(self,x):
    if self.front == -1:
      self.front = 0
    self.q.append(x)

  def pop(self):
    if len(self.q) == 0:
      return -1
    x = self.q[self.front]
    self.front +=1
    if len(self.q) == self.front:
      self.front = -1
    return x

  def getFront(self):
    if len(self.q) == 0:
      return -1
    return self.q[self.front]

  def size(self):
    if self.front == -1:
      return 0
    return len(self.q) - self.front


q = Queue()
visited = [False]*n
ans = []

q.push(0)
visited[0] = True
ans.append(0)

while q.size()>0:
  front = q.pop()
  for i in adjecency_list[front]:
    if not visited[i]:
        q.push(i)
        visited[i] = True
        ans.append(i)
print(ans)

```

```python
#DFS using stack
class Stack:
  def __init__(self):
    self.st = []

  def push(self,x):
    self.st.append(x)

  def pop(self):
    if len(self.st) == 0:
      return -1
    x = self.st[-1]
    self.st.pop()
    return x

  def top(self):
    if len(self.q) == 0:
      return -1
    return self.st[-1]

  def size(self):
    return len(self.st)


st = Stack()
visited = [False]*n
ans = []

st.push(0)
visited[0] = True
ans.append(0)

while st.size()>0:
  top = st.pop()
  for i in adjecency_list[top]:
    if not visited[i]:
        st.push(i)
        visited[i] = True
        ans.append(i)
print(ans)
```

```python
# Detect cycle in an undirected graph
visited = [False]*n

def dfs(i ,parent, adjecency_list , visited):
  visited[i] = True
  for x in adjecency_list[i]:
    if x == parent:
      continue
    if visited[x]:
        return True
    if dfs(x ,i, adjecency_list , visited):
      return True
  return False
dfs(0 ,-1,  adjecency_list , visited)

```

```python
from heapq import heappush,heappop
n =  6
s = 0
edges = [(0,1,4),(0,2,4),(1,2,2),(2,3,3),(2,4,1),(2,5,6),(3,5,2),(4,5,3)]
# dijkatra's algorithm
adjecency_list = []
for i in range(n):
  adjecency_list.append([])
for edge in edges:
  x = edge[0]
  y = edge[1]
  w = edge[2]
  adjecency_list[x].append([y,w])
  adjecency_list[y].append([x,w])
visited = [False]*n
heap = []
dist = [float("inf")]*n
visited[s] = True
dist[s] = 0
heappush(heap,(dist[s],s))
for i in range(n):
  print(i,"-->",adjecency_list[i])
```

```python
# Prim's Algorithm
G= [[0,9,75,0,0],
    [9,0,95,19,42],
    [75,95,0,51,66],
    [0,19,51,0,31],
    [0,42,66,31,0]]
n = 5
visited = [False]*n
visited[0] = True
ans = 0
for _ in range(n-1):
  x= -1
  y = -1
  mn = float("inf")
  for i in range(n):
    for j in range(n):
      if G[i][j]!=0 and visited[i] and not visited[j]:
        if G[i][j]<mn:
          mn = G[i][j]
          x = i
          y = j
  visited[y] = True
  ans+=mn
print(ans)
```

# Dynamic Programming

```python
# fibonacci problem using memoization
class Solution:
    def rec(self , n , dp):
        if n==0 or n==1:
            return n
        if dp[n]!=-1:
            return dp[n]
        dp[n] = self.rec(n-1,dp) + self.rec(n-2,dp)
        return dp[n]
    def fib(self, n: int) -> int:
        dp = [-1]*(n+1)
        return self.rec(n,dp)
```

```python
# fibonacci problem using tabulation
class Solution:
    def fib(self, n: int) -> int:
        if n == 1 or n == 0:
            return n
        dp = [-1]*(n+1)
        dp[0]=0
        dp[1]=1
        for i in range(2 , n+1):
            dp[i] = dp[i-1] + dp[i-2]
        return dp[n]
```
